```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import math 
```

```python
def sigmoid_function(z):
    if z < -500:
        return 0.0
    return 1 / (1 + math.exp(-z))

def function(x,w,b):
    return w*x+b

# print(sigmoid_function(function(1,0.1,0.2)))
```

```python
#now making a plot over this thing
x_inputs=[]
for i in range(-10,10):
    x_inputs.append(i)
    
y_outputs = []
for x in x_inputs:
    y_outputs.append(sigmoid_function(x))

plt.plot(x_inputs, y_outputs)
plt.show()
```
