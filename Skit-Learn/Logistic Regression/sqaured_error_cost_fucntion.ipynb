```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import math 
```

```python
X = np.array([0.5, 1.0, 1.5, 2.0, 2.5, 3.0, 3.5, 4.0, 10.0])
y = np.array([0,   0,   0,   1,   0,   1,   1,   1,   1])

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

weights = np.linspace(-5, 5, 200)

mse_costs = []
log_loss_costs = []

for w in weights:
    y_pred = sigmoid(w * X)
    mse = np.mean((y_pred - y)**2)
    mse_costs.append(mse)
    
    y_pred_clipped = np.clip(y_pred, 1e-15, 1 - 1e-15)
    log_loss = -np.mean(y * np.log(y_pred_clipped) + (1 - y) * np.log(1 - y_pred_clipped))
    log_loss_costs.append(log_loss)


fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))

ax1.plot(weights, mse_costs, color='red', linewidth=2)
ax1.set_title('Real MSE on Logistic Reg.\n(Non-Convex / Plateaus)', fontsize=14, fontweight='bold')
ax1.set_xlabel('Weight (w)', fontsize=12)
ax1.set_ylabel('Cost (MSE)', fontsize=12)
ax1.grid(True, linestyle='--', alpha=0.6)   

ax2.plot(weights, log_loss_costs, color='blue', linewidth=2)
ax2.set_title('Real Log Loss on Logistic Reg.\n(Strictly Convex)', fontsize=14, fontweight='bold')
ax2.set_xlabel('Weight (w)', fontsize=12)
ax2.set_ylabel('Cost (Log Loss)', fontsize=12)
ax2.grid(True, linestyle='--', alpha=0.6)

plt.tight_layout()
plt.show()
```
