```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import time as t
```

```python
data=pd.read_csv('./housedata.csv')
display(data)
data.shape
```

```python
w = [1, 2]
b = 1

m = len(data)
n = data.shape[1] - 1

start = t.perf_counter()

for i in range(m):
    y = 0

    for j in range(n):
        y += data.iloc[i][j] * w[j]

    y += b

end = t.perf_counter()

# now with vectorization
# we just have to take simple dot product
X = data[["size", "bedrooms"]].values
start_new=t.perf_counter()
for i in range(m):
    prediction=np.dot(X[i],w)+b

end_new=t.perf_counter()
print("new time",end_new-start_new)


print("time required was", end - start)
```
