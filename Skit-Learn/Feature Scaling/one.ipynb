```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
data=pd.read_csv('./house_data.csv')
display(data)
```

```python

#calculations for b
def gradient_descnt(w,b,learning_rate,X,y):
    m = len(X)
    while True:

        dj_db = 0

        for i in range(m):
            prediction = np.dot(X[i], w) + b
            error = prediction - y[i]
            dj_db += error

        dj_db = dj_db / m

        b_new = b - learning_rate * dj_db

        if abs(b_new - b) < 1e-6:
            b = b_new
            break

        b = b_new
    
    #calculations for w
    while (True):
        dj_dw=0
        for i in range(m):
            prediction = np.dot(X[i], w) + b
            error = (prediction - y[i])*X[i]
            dj_db += error
        
        dj_dw=dj_dw/m
        w_new=w-learning_rate*dj_dw
        
        if abs(w_new - w) < 1e-6:
                w = w_new
                break
        
        w = w_new

    return w,b
```

```python

temp_X = data[['size', 'bedrooms']].copy()
y = data['price'].values


w = np.array([1.0, 2.0])
b = 1.0
learning_rate = 0.1


temp_X['size'] /= 2000
temp_X['bedrooms'] /= 4


X = temp_X.values

(w, b) = gradient_descnt(w, b, learning_rate, X, y)

print("Updated weights:", w)
print("Updated bias:", b)
```
