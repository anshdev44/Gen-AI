```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data=pd.read_csv("./exam_scores.csv")
# display(data)
# for the total number of rows which is m
m=len(data)
print(m)
# data.shape[0]
x=data['Hours']
print(x)
y_actual=data['Score']
# print(y_actual)
```

```python
#now based of the w and b values we wil implement th cost funtion
w = 1.1
b = 0

y_predicted = w*x +b

plt.scatter(x, y_actual)

plt.xlabel("Hours Studied")
plt.ylabel("Exam Score")
plt.title("Training Data")

plt.show()


# print(y_predicted)
```

```python
#now we are goonna implement the cost function
error=0
for i in range(len(data)):
    error+=(y_actual[i]-y_predicted[i])**2

j=error/(2*m)
print(j)
    
plt.scatter(x, y_actual)

plt.plot(x, y_predicted)

plt.xlabel("Hours Studied")
plt.ylabel("Exam Score")
plt.title("Prediction vs Actual")

plt.show()
    
```
