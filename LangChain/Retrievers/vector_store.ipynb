```python
from langchain_community.vectorstores import Chroma
from langchain_google_genai import GoogleGenerativeAI,GoogleGenerativeAIEmbeddings
from langchain_core.documents import Document
from dotenv import load_dotenv
load_dotenv()
 
```

```python
documents = [
    Document(page_content="LangChain helps developers build LLM applications easily."),
    Document(page_content="Chroma is a vector database optimized for LLM-based search."),
    Document(page_content="Embeddings convert text into high-dimensional vectors."),
    Document(page_content="OpenAI provides powerful embedding models."),
]
```

```python
# Create an embedding model that converts text into numerical vectors
embedding = GoogleGenerativeAIEmbeddings(model="gemini-embedding-2")

# Create a Chroma vector store: embed the documents and store their vectors + text
vector_store = Chroma.from_documents(
    documents=documents, embedding=embedding, collection_name="my_collection"
)

# The question we want to search for in our stored documents
query = "what is chroma vector database"

# Turn the vector store into a retriever and ask it to return the top 2 most relevant documents
retriever = vector_store.as_retriever(search_kwargs={"k": 2})

# Run the query through the retriever and get the 2 most relevant documents
results = retriever.invoke(query)

# Loop through each retrieved document so we can display the results
for i, doc in enumerate(results):

    # Print the result number (i starts from 0, so we use i + 1)
    print(f"\n--- Result {i+1} ---")

    # Print the actual text/content of the retrieved document
    print(doc.page_content)
```
