```python
import numpy as np
```

```python
a = np.arange(10,0,-1).reshape(2,5)
print(a)
# print(np.sort(a))
# print(a[1])
print(a[:,1])
```

```python
a=np.array([[1,2,3],[4,5,6]])
print(a[a%2==0])
mask=a%2==0
print(a[mask])
```

```python
#using the where clause
a=np.array([[1,2,3],[4,5,6]])
where_result=np.where(a%2==0,1,0)
print(where_result)
```

```python
a=np.array([[1,2,3],[4,5,6]])
b=np.array([[7,8,9]])
print(np.concatenate((a,b)))
```

```python
a=np.array([[1,2,3],[4,5,6]])
b=np.array([[7,8,9],[10,11,12]])
c=np.hstack((a,b))
print(c)
```
