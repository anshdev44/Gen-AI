```python
a=np.arange(1,21,1).reshape((4,5))
print(a)
#using slicing to get result
print(a[1:5,2:4])
# b=[0,1,2,3,4]
# print(b[2:5])

#reversing a matrix
b = np.empty_like(a)

i = 3

for k in range(4):
    j = 4
    for l in range(5):
        b[k, l] = a[i, j]
        j -= 1
    i -= 1
print(b)


#now replace 3 with -1
c=np.where(a%3==0,3,a)
print(c)
```
