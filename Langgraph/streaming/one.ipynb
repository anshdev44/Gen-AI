```python
from langchain_google_genai import ChatGoogleGenerativeAI
from typing import TypedDict,Annotated,Literal
from langgraph.graph import StateGraph, MessagesState, START, END
#this imports the checkpoiters and save the data into the RAM
from langgraph.checkpoint.memory import InMemorySaver
from dotenv import load_dotenv
load_dotenv()
```

```python
llm=ChatGoogleGenerativeAI(model="gemini-3.7-flash")
```

```python
class EssayState(TypedDict):
    topic: str
    essay: str
    
def generate_essay(state:EssayState):
   prompt=f'Generate a 100 word essay on the topic->{state["topic"]}'
   result=llm.invoke(prompt).content
   
   return {'essay':result}



graph = StateGraph(EssayState)
graph.add_node('generate_essay', generate_essay)

graph.add_edge(START, 'generate_essay')
graph.add_edge('generate_essay', END)

checkpointer = InMemorySaver()

chatbot=graph.compile(checkpointer=checkpointer)

config = {"configurable": {"thread_id": "1"}}

for chunk in chatbot.stream(
    {"topic": "Artificial Intelligence"},
    config=config,
    stream_mode="updates"
):
    print(chunk)
```
