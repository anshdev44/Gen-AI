```python
coffee=pd.read_csv('./warmup-data/coffee.csv')
coffee['price']=2.99
mask = (coffee["Coffee Type"] == "Espresso") & (coffee["Day"] == "Monday")
coffee['new_price']=np.where(mask ,7.99,2.99)
coffee = coffee.drop(columns=["price"])
coffee['revenue']=coffee['Units Sold']*coffee['new_price']
coffee.rename(columns={'new_price':'price'},inplace=True)
# display(coffee)

#using apply function to make categories
#this apply fucntion takes every coloumn from the table and apply this fucntion
# bios=pd.read_csv("./data/bios.csv")
#adding to a new coloumn of date and time 
bios['born-datetime'] = pd.to_datetime(
    bios['born_date']
)
bios['height-category'] = bios['height_cm'].apply(
    lambda x: 'Short' if x < 165 else ('Average' if x < 185 else 'Tall')
)
#we can also make a cu
display(bios)
```

```python
#we can also make a customized function 
def categorize_weight(row):
    if row['weight_kg']<60:
        return 'Light-Weight'
    elif row['weight_kg']<75:
        return 'Fit'
    else:
        return 'Heavy-Weight'
    
bios['Weight-Category']=bios.apply(categorize_weight,axis=1)

display(bios)
        
```

```python
coffee=pd.read_csv('./warmup-data/coffee.csv')
#replacing some values with nan
coffee.loc[[0,1],"Units Sold"]=np.nan
new_coffee=coffee.fillna(coffee['Units Sold'].mean())
# display(new_coffee)

#fills the estimated value if nan is between any two null values 
coffee['Units Sold']=coffee['Units Sold'].interpolate()

# Drop rows where 'Units Sold' is missing
coffee.dropna(subset=["Units Sold"], inplace=True)
# subset=["Units Sold"] -> Check NaN only in this column
# inplace=True -> Modify the original DataFrame
coffee
```
